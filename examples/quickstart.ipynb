# PathéResistance · offline quickstart
Install the package into your selected Python environment first: `python -m pip install .` from the repository root. Use `python -m pip install '.[notebook]'` if you need a Jupyter kernel.

This example uses six **synthetic** nodes and a tiny synthetic tree raster, not real imagery or AI output. It makes no network requests. For a real project, use a JSON config with your boundary or GraphML and georeferenced predictions; see the README.

```python
import json
import tempfile
from pathlib import Path

from pathe_resistance.demo import run_demo

work = Path(tempfile.mkdtemp(prefix="pathe-resistance-"))
manifest = run_demo(work / "demo")
results = work / "demo/results"
print("Outputs:", results)
print(manifest["graph"])
```

```python
comparison = json.loads((results / "route_comparison.json").read_text())[0]
for label in ["baseline", "resistance"]:
    route = comparison[label]
    print(
        label,
        "nodes:",
        route["nodes"],
        "meters:",
        route["distance_m"],
        "modeled walking seconds:",
        round(route["walking_seconds"], 1),
    )
print("Baseline resistance:", round(comparison["baseline_resistance_s"], 1))
print("Selected resistance:", round(comparison["resistance"]["objective_cost"], 1))
```

To adapt this, replace the synthetic inputs with your own GraphML and predicted raster. `tree_cover` is sampled beside paths with explicit pixel coverage; unknown predictions stay unknown. Coefficients are declared preferences. The engine does not train a comfort model or validate real crossings.

Optional inspection map: install the `map` extra and set `"map": true` in a config. See `examples/area.json` and `examples/shade_priority.json`.
